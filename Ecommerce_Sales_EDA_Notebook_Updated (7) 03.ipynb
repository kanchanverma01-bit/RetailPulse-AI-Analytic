### 4. Products Likely to Run Out of Stock (Inventory-Based Stock-Out Risk)

**Note:** The raw sales file does not include real inventory fields, so we build an **Inventory table** here using realistic, reproducible values (instead of the earlier demand-velocity proxy). For every product we add:

- **Unit Price** – taken directly from the sales data (`Unit Price`)
- **Current_Stock_On_Hand** – units currently available in the warehouse
- **Reorder_Level** – the minimum stock threshold at which a reorder should be triggered
- **Last_Restock_Date** – the date stock was last replenished
- **Avg_Daily_Sales** – actual daily sales velocity calculated from `Order Date` + `Quantity`
- **Days_Of_Stock_Left** – `Current_Stock_On_Hand / Avg_Daily_Sales`
- **Stock_Status** – `Reorder Now` / `Low Stock` / `OK` based on stock vs reorder level and days left

> In production you would pull `Current_Stock_On_Hand`, `Reorder_Level`, and `Last_Restock_Date` straight from your inventory/WMS system. Since this dataset doesn't have one, the values below are generated with a fixed random seed so results are reproducible — replace the `np.random` lines with a merge against your real inventory table when available.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Ensure df is loaded even if this cell is run standalone / after kernel restart
if 'df' not in globals():
    df = pd.read_csv("Ecommerce_Sales_Data_2024_2025.csv")

np.random.seed(42)

df['Order Date'] = pd.to_datetime(df['Order Date'])
df['YearMonth'] = df['Order Date'].dt.to_period('M')

# ---- 1) Real sales velocity per product (from actual order data) ----
total_qty_per_product = df.groupby('Product Name')['Quantity'].sum()
date_range_days = (df['Order Date'].max() - df['Order Date'].min()).days
date_range_days = max(date_range_days, 1)

avg_unit_price = df.groupby('Product Name')['Unit Price'].mean()

inventory_df = pd.DataFrame({
    'Product Name': total_qty_per_product.index,
    'Unit Price': avg_unit_price.reindex(total_qty_per_product.index).round(2).values,
    'Avg_Daily_Sales': (total_qty_per_product / date_range_days).round(2).values
})

# ---- 2) Inventory fields (Current Stock, Reorder Level, Last Restock Date) ----
# Replace this block with a real merge to your inventory/WMS export when available.
n = len(inventory_df)
inventory_df['Current_Stock_On_Hand'] = np.random.randint(5, 200, size=n)
inventory_df['Reorder_Level'] = (inventory_df['Avg_Daily_Sales'] * np.random.randint(7, 15, size=n)).round(0)
inventory_df['Last_Restock_Date'] = pd.to_datetime(
    df['Order Date'].max() - pd.to_timedelta(np.random.randint(1, 60, size=n), unit='D')
)

# ---- 3) Derived stock-out risk metrics ----
inventory_df['Days_Of_Stock_Left'] = (
    inventory_df['Current_Stock_On_Hand'] / inventory_df['Avg_Daily_Sales'].replace(0, np.nan)
).round(1)

def stock_status(row):
    if row['Current_Stock_On_Hand'] <= row['Reorder_Level']:
        return 'Reorder Now'
    elif pd.notna(row['Days_Of_Stock_Left']) and row['Days_Of_Stock_Left'] <= 7:
        return 'Low Stock'
    else:
        return 'OK'

inventory_df['Stock_Status'] = inventory_df.apply(stock_status, axis=1)

# ---- 4) Products at high risk of running out ----
stockout_risk = inventory_df[inventory_df['Stock_Status'].isin(['Reorder Now', 'Low Stock'])] \
                    .sort_values('Days_Of_Stock_Left')

cols_to_show = ['Product Name', 'Unit Price', 'Current_Stock_On_Hand', 'Reorder_Level',
                 'Avg_Daily_Sales', 'Days_Of_Stock_Left', 'Last_Restock_Date', 'Stock_Status']

print(f'{len(stockout_risk)} products flagged as at risk of running out of stock:\n')
print(stockout_risk[cols_to_show].head(15))

plt.figure(figsize=(10,6))
top15 = stockout_risk.head(15)
sns.barplot(x=top15['Days_Of_Stock_Left'], y=top15['Product Name'],
            hue=top15['Stock_Status'], dodge=False, palette='flare')
plt.title('Top 15 Products at Risk of Stock-Out (Days of Stock Left)')
plt.xlabel('Days of Stock Left')
plt.ylabel('Product Name')
plt.legend(title='Stock Status')
plt.tight_layout()
plt.show()

inventory_df.head()
```

```python
# ---- Full Inventory Table (all required columns together) ----
# Build inventory_df here too, in case this cell is run standalone / after kernel restart
if 'inventory_df' not in globals():
    import numpy as np
    import pandas as pd

    if 'df' not in globals():
        df = pd.read_csv("Ecommerce_Sales_Data_2024_2025.csv")
        df['Order Date'] = pd.to_datetime(df['Order Date'])

    np.random.seed(42)

    total_qty_per_product = df.groupby('Product Name')['Quantity'].sum()
    date_range_days = max((df['Order Date'].max() - df['Order Date'].min()).days, 1)
    avg_unit_price = df.groupby('Product Name')['Unit Price'].mean()

    inventory_df = pd.DataFrame({
        'Product Name': total_qty_per_product.index,
        'Unit Price': avg_unit_price.reindex(total_qty_per_product.index).round(2).values,
        'Avg_Daily_Sales': (total_qty_per_product / date_range_days).round(2).values
    })

    n = len(inventory_df)
    inventory_df['Current_Stock_On_Hand'] = np.random.randint(5, 200, size=n)
    inventory_df['Reorder_Level'] = (inventory_df['Avg_Daily_Sales'] * np.random.randint(7, 15, size=n)).round(0)
    inventory_df['Last_Restock_Date'] = pd.to_datetime(
        df['Order Date'].max() - pd.to_timedelta(np.random.randint(1, 60, size=n), unit='D')
    )
    inventory_df['Days_Of_Stock_Left'] = (
        inventory_df['Current_Stock_On_Hand'] / inventory_df['Avg_Daily_Sales'].replace(0, np.nan)
    ).round(1)

    def stock_status(row):
        if row['Current_Stock_On_Hand'] <= row['Reorder_Level']:
            return 'Reorder Now'
        elif pd.notna(row['Days_Of_Stock_Left']) and row['Days_Of_Stock_Left'] <= 7:
            return 'Low Stock'
        else:
            return 'OK'

    inventory_df['Stock_Status'] = inventory_df.apply(stock_status, axis=1)

inventory_full = inventory_df[['Product Name', 'Unit Price', 'Current_Stock_On_Hand',
                                'Reorder_Level', 'Avg_Daily_Sales', 'Days_Of_Stock_Left',
                                'Last_Restock_Date', 'Stock_Status']].sort_values('Days_Of_Stock_Left')

print(f'Full Inventory Table ({len(inventory_full)} products):\n')
inventory_full
```

### 5. Expected Sales Forecast for Next Month
A simple **linear trend forecast** is built on historical monthly total sales to estimate next month's expected sales. For production use, a more robust model (e.g., ARIMA, Prophet, or exponential smoothing) is recommended.

```python
# Next Month Sales Forecast using NumPy linear trend (no external ML library needed)
monthly_total_sales = df.groupby('YearMonth')['Sales'].sum().sort_index()
monthly_total_sales.index = monthly_total_sales.index.to_timestamp()

ts = monthly_total_sales.reset_index()
ts.columns = ['Month','Sales']
ts['t'] = np.arange(len(ts))

# Fit a simple linear trend: Sales = slope * t + intercept
slope, intercept = np.polyfit(ts['t'], ts['Sales'], 1)

next_t = len(ts)
next_month_forecast = slope * next_t + intercept
next_month_label = (ts['Month'].max() + pd.DateOffset(months=1)).strftime('%B %Y')

print(f'Expected Sales for {next_month_label}: {next_month_forecast:,.2f}')

trend_line = slope * ts['t'] + intercept

plt.figure(figsize=(10,5))
plt.plot(ts['Month'], ts['Sales'], marker='o', label='Historical Sales')
plt.plot(pd.to_datetime(next_month_label), next_month_forecast, 'ro', markersize=10, label='Forecast (Next Month)')
plt.plot(ts['Month'], trend_line, linestyle='--', color='gray', label='Trend Line')
plt.title('Monthly Sales Trend & Next Month Forecast')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
